# Hugging Face Text-to-Speech — Kokoro-82M

Synthesize speech from text with **[hexgrad/Kokoro-82M](https://huggingface.co/hexgrad/Kokoro-82M)** using the `kokoro` library.

Requires **espeak-ng** on the system (`brew install espeak-ng` on macOS).


## Load pipeline

`lang_code='a'` is American English. Weights download from Hugging Face on first run.


```python
import os
from pathlib import Path

os.environ["HF_HOME"] = os.path.expanduser("~/ml-models")

# importing the constants after setting the env var.
from huggingface_hub.constants import HF_HOME, HF_HUB_CACHE
print("HF_HOME:", HF_HOME)
print("hub cache:", HF_HUB_CACHE)
```

```python
from IPython.display import Audio, display
from kokoro import KPipeline
import soundfile as sf

pipeline = KPipeline(lang_code="a", repo_id="hexgrad/Kokoro-82M")
OUTPUT_DIR = Path("output")
OUTPUT_DIR.mkdir(exist_ok=True)
```

## Synthesize

Change `voice` to try other speakers (see [VOICES.md](https://huggingface.co/hexgrad/Kokoro-82M/blob/main/VOICES.md)). Sample rate is 24 kHz.


```python
text = """
[Kokoro](/kˈOkəɹO/) is an open-weight TTS model with 82 million parameters.
Despite its lightweight architecture, it delivers comparable quality to larger models
while being significantly faster and more cost-efficient.
"""

VOICE = "af_heart"
SAMPLE_RATE = 24_000

generator = pipeline(text, voice=VOICE, speed=1, split_pattern=r"\n+")

for i, (gs, ps, audio) in enumerate(generator):
    print(i, gs)
    print("phonemes:", ps)
    display(Audio(data=audio, rate=SAMPLE_RATE, autoplay=(i == 0)))
    out_path = OUTPUT_DIR / f"{i}_{VOICE}.wav"
    sf.write(out_path, audio, SAMPLE_RATE)
    print("saved:", out_path)
```

## Try your own text


```python
custom_text = "Hello from Hugging Face. This is Kokoro text to speech running locally."

generator = pipeline(custom_text, voice=VOICE, speed=1)
for i, (gs, ps, audio) in enumerate(generator):
    display(Audio(data=audio, rate=SAMPLE_RATE, autoplay=True))
    out_path = OUTPUT_DIR / f"custom_{VOICE}.wav"
    sf.write(out_path, audio, SAMPLE_RATE)
    print("saved:", out_path)
```
